### Why Environment Validation Matters

Machine learning projects depend on specific Python environments and software packages.

Before building models, we must confirm that our notebook uses the correct interpreter and dependencies.

This helps prevent errors, improves reproducibility, and makes the project easier for other developers to understand.

```python

import sys

print("Python version:", sys.version.split()[0])
print("Python executable:", sys.executable)
print("Virtual environment active:", sys.prefix != sys.base_prefix)
```

```python

import pytest
import jupyterlab
import ipykernel

print("pytest:", pytest.__version__)
print("JupyterLab:", jupyterlab.__version__)
print("ipykernel:", ipykernel.__version__)

print("Package validation: PASS")
```

```python

import sys
from pathlib import Path

project_name = "comicvision-ai"

python_path = Path(sys.executable)

assert sys.version_info >= (3, 11), "Python 3.11+ required"
assert sys.prefix != sys.base_prefix, "Virtual environment required"
assert project_name in str(python_path).lower(), "Wrong project environment"

print("All environment assertions passed!")
```
